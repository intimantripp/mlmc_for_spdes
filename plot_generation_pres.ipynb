```python
L = 1.0
nf = 4
nc = 2
xf = np.linspace(0, L, nf+1)   # fine grid including endpoints
xc = np.linspace(0, L, nc+1)   # coarse grid

ax = plt.gca()

# fine control volumes
ax.vlines([1/8, 3/8, 5/8, 7/8], ymin=4, ymax=5, color='black', linestyle='--')
ax.hlines([5,4,], xmin=0, xmax=1, color='black', linestyle='--')
ax.text(0.02, 4.4, r"$\Delta W_0$", color='black')
ax.text(0.22, 4.4, r"$\Delta W_1$", color='black')
ax.text(0.47, 4.4, r"$\Delta W_2$", color='black')
ax.text(0.72, 4.4, r"$\Delta W_3$", color='black')
ax.text(0.92, 4.4, r"$\Delta W_4$", color='black')

# fine basis functions
fine_h = 0.25
fine_phi = lambda x_i, x: np.maximum(0, 1 - abs(x - x_i) / fine_h)
fine_nodes = [0.0, 0.25, 0.5, 0.75, 1.0]
hat_height = 1
for x_i in fine_nodes:
        x0, x1 = x_i - fine_h, x_i + fine_h
        x = np.linspace(max(0, x0), min(1, x1), 100)
        y = 2 + hat_height * fine_phi(x, x_i)
        ax.plot(x, y, color='tab:blue', linestyle='--', lw=1.5)
ax.hlines(2, xmin=0, xmax=1, color='black', linestyle='-')
ax.text(-0.01, 1.8, r"$\phi_0$", color='black')
ax.text(0.235, 1.8, r"$\phi_1$", color='black')
ax.text(0.485, 1.8, r"$\phi_2$", color='black')
ax.text(0.735, 1.8, r"$\phi_3$", color='black')
ax.text(0.985, 1.8, r"$\phi_4$", color='black')

# coarse basis functions
coarse_h = 0.5
coarse_phi = lambda x_i, x: np.maximum(0, 1 - abs(x - x_i) / coarse_h)
fine_nodes = [0.0, 0.5, 1.0]
hat_height = 1
for x_i in fine_nodes:
        x0, x1 = x_i - coarse_h, x_i + coarse_h
        x = np.linspace(max(0, x0), min(1, x1), 100)
        y = 0 + hat_height * coarse_phi(x, x_i)
        ax.plot(x, y, color='tab:red', linestyle='--', lw=1.5)
ax.hlines(0, xmin=0, xmax=1, color='black', linestyle='-')
ax.text(-0.01, -0.2, r"$\phi_0$", color='black')
ax.text(0.485, -0.2, r"$\phi_1$", color='black')
ax.text(0.985, -0.2, r"$\phi_2$", color='black')

arrows = [
        ([0.25, 3.8], [0.25, 3.2]),
        ([0.5, 3.8], [0.5, 3.2]),
        ([0.75, 3.8], [0.75, 3.2]),
        ([0.25, 1.6], [0.25, 1.0]),
        ([0.5, 1.6], [0.5, 1.0]),
        ([0.75, 1.6], [0.75, 1.0]),
]
for start, end in arrows:
    ax.annotate("",
                xy=end, xycoords='data',
                xytext=start, textcoords='data',
                arrowprops=dict(arrowstyle="->", color="black", lw=1.0),
                label='')
                
ax.axis('off')
plt.savefig("presentation/presentation_graphics/fe.png", dpi=300, bbox_inches="tight")
plt.show()
```
